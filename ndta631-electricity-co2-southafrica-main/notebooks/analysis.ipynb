
Section 1: Data Preparation

```python
import pandas as pd
import numpy as np
```

```python
elec_raw = pd.read_csv("../data/WB_WDI_EG_ELC_ACCS_ZS (1).csv")
elec_raw


```

```python
south_africa = elec_raw[elec_raw["REF_AREA"] == "ZAF"]
south_africa










```

```python
south_africa_clean = south_africa[["TIME_PERIOD", "OBS_VALUE"]]
south_africa_clean
```

```python
south_africa_clean.columns
```

```python
south_africa_clean = south_africa_clean.rename(columns={"TIME_PERIOD": "Year", "OBS_VALUE": "Electricity_Access_Pct"})
south_africa_clean.columns
```

```python
south_africa_clean = south_africa_clean.sort_values("Year")
south_africa_clean
```

```python
co2_raw = pd.read_csv("../data/WB_ESG_EN_ATM_CO2E_PC.csv")
co2_raw
```

```python
south_africa_co2 = co2_raw[co2_raw["REF_AREA"] == "ZAF"]
south_africa_co2
```

```python
south_africa_co2_clean = south_africa_co2[["TIME_PERIOD", "OBS_VALUE"]]
south_africa_co2_clean = south_africa_co2_clean.rename(columns={"TIME_PERIOD": "Year", "OBS_VALUE": "CO2_Emissions_PerCapita"})
south_africa_co2_clean = south_africa_co2_clean.sort_values("Year")
south_africa_co2_clean
```

```python
combined = pd.merge(south_africa_clean, south_africa_co2_clean, on="Year", how="inner")
combined
```

```python
combined.isna().sum()
```





SECTION 2: Numerical Analysis

2.1 Reshaping and period averages

```python
years = combined["Year"].to_numpy()
elec = combined["Electricity_Access_Pct"].to_numpy()
co2 = combined["CO2_Emissions_PerCapita"].to_numpy()

print(years)
print(elec)
print(co2)
```

```python
print(years.shape)
print(elec.shape)
print(co2.shape)
```

```python
n_years = len(years)
print("Total years:", n_years)
```

```python
elec_grid = elec.reshape(5, 5)
print(elec_grid)
```

```python
elec_period_avg = elec_grid.mean(axis=1)

print(elec_period_avg)
```

```python
co2_grid = co2.reshape(5, 5)
co2_period_avg = co2_grid.mean(axis=1)
print(co2_grid)
print(co2_period_avg)
```



2.2 Year-on-year change 

```python
elec_yoy = np.diff(elec)
co2_yoy = np.diff(co2)

print(elec_yoy)
print(co2_yoy)
```



2.3 Rolling average

```python
print("Average yearly electricity gain:", elec_yoy.mean())
print("Average yearly CO2 change:", co2_yoy.mean())
```



2.4 Correlation

```python
corr_matrix = np.corrcoef(elec, co2)
print(corr_matrix)

r = corr_matrix[0, 1]
print("Correlation coefficient:", r)
```



2.5 Trend line 

```python
elec_slope, elec_intercept = np.polyfit(years, elec, 1)
co2_slope, co2_intercept = np.polyfit(years, co2, 1)

print("Electricity trend (per year):", elec_slope)
print("CO2 trend (per year):", co2_slope)
```



2.6 Normalisation (z-scores)

```python
elec_z = (elec - elec.mean()) / elec.std()
co2_z = (co2 - co2.mean()) / co2.std()

print(elec_z)
print(co2_z)
```




Section 3 - Visualisation

```python



import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(10, 5))
plt.plot(years, elec, marker='o', color="steelblue" )
plt.title("South Africa: Electricity Access Over Time (1996-2020)")
plt.xlabel("Year")
plt.ylabel("Electricity Access (%)")
plt.grid(True)
plt.show()
```

Trend: Electricity access in South Africa rose almost continuously from 57.6% in 1996
to 90.0% in 2020 — a gain of roughly 32 percentage points over 25 years. The climb is steady
rather than sudden, consistent with a long-running national electrification programme rather
than a single policy event.

```python





plt.figure(figsize=(10, 5))
plt.plot(years, co2, marker="o", color="darkorange")
plt.title("South Africa: CO2 Emissions Per Capita Over Time (1996-2020)")
plt.xlabel("Year")
plt.ylabel("CO2 Emissions (metric tons per capita)")
plt.grid(True)
plt.show()
```

Trend: CO2 emissions per capita rise more gently, peak around 2008 (~8.4 t/capita), then
decline through the 2010s down to around 6.5 t/capita by 2020. Unlike electricity access,
this is not a one-directional trend — it rises, plateaus, then falls.

```python
fig, ax1 = plt.subplots(figsize=(10, 5))

ax1.plot(years, elec, marker="o", color="steelblue", label="Electricity Access (%)")
ax1.set_xlabel("Year")
ax1.set_ylabel("Electricity Access (%)", color="steelblue")
ax1.tick_params(axis="y", labelcolor="steelblue")

ax2 = ax1.twinx()
ax2.plot(years, co2, marker="s", color="darkorange", label="CO2 Emissions (t/capita)")
ax2.set_ylabel("CO2 Emissions (metric tons per capita)", color="darkorange")
ax2.tick_params(axis="y", labelcolor="darkorange")

plt.title("South Africa: Electricity Access vs CO2 Emissions (1996-2020)")
fig.tight_layout()
plt.show()
```

Trend: The two lines move together in the earlier years but visibly diverge after around
2010 — electricity access keeps climbing while CO2 emissions per capita level off and decline.
This is the clearest single chart illustrating the assignment's core story.

```python




period_labels = ["1996-2000", "2001-2005", "2006-2010", "2011-2015", "2016-2020"]

plt.figure(figsize=(10, 5))
plt.bar(period_labels, elec_period_avg, color="steelblue")
plt.title("South Africa: Average Electricity Access by 5-Year Period")
plt.xlabel("Period")
plt.ylabel("Average Electricity Access (%)")
plt.show()
```

Trend: Electricity access grows in every successive 5-year period, but the size of the
gain shrinks over time (large jumps in the early periods, smaller gains later) as access
approaches saturation near 90%.

```python



plt.figure(figsize=(8, 6))
plt.scatter(elec, co2, color="purple")
plt.title("South Africa: Electricity Access vs CO2 Emissions (each dot = one year)")
plt.xlabel("Electricity Access (%)")
plt.ylabel("CO2 Emissions (metric tons per capita)")
plt.show()
```

Trend: The dots trend broadly upward and to the right, consistent with the positive
correlation (r = 0.73) calculated in Section 2 — but the spread is wide, meaning the
relationship is real but not tight or perfectly predictable.

```python


plt.figure(figsize=(8, 5))
plt.hist(co2, bins=8, color="darkorange", edgecolor="black")
plt.title("South Africa: Distribution of CO2 Emissions per Capita (1996-2020)")
plt.xlabel("CO2 Emissions (metric tons per capita)")
plt.ylabel("Number of Years")
plt.show()
```

Trend: CO2 emissions per capita cluster mostly between 6.5 and 8 tons, with no extreme
outliers — the country's per-capita carbon footprint has stayed within a fairly narrow band
over the 25-year period.

```python


plt.figure(figsize=(6, 6))
plt.boxplot([elec, co2], tick_labels=["Electricity Access (%)", "CO2 (t/capita)"])
plt.title("Spread of Electricity Access vs CO2 Emissions")
plt.show()
```

Trend: Electricity access has a much wider spread (min ~58%, max 90%) than CO2 emissions,
which cluster tightly around 7 tons with a narrow box — visually confirming that electricity
access moved far more than emissions did over the same period.




Section 4: Database Intergration

4.1 Connected to a database

```python
import sqlite3

conn = sqlite3.connect("south_africa_data.db")
cursor = conn.cursor()
```

4.2 Created a table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS electricity_co2 (
    Year INTEGER PRIMARY KEY,
    Electricity_Access_Pct REAL,
    CO2_Emissions_PerCapita REAL
)
""")
conn.commit()
```

4.3 Loaded the cleaned dataset into the database

```python
combined.to_sql("electricity_co2", conn, if_exists="replace", index=False)
```

4.4 Queried it back

```python
result = cursor.execute("SELECT * FROM electricity_co2 ORDER BY Year").fetchall()
for row in result:
    print(row)
```

4.5 Updated a record safely

```python
new_value = 90.0
year_to_update = 2020

cursor.execute(
    "UPDATE electricity_co2 SET Electricity_Access_Pct = ? WHERE Year = ?",
    (new_value, year_to_update)
)
conn.commit()

print(f"Updated {cursor.rowcount} row(s)")
```

4.6 Deleted a record safely

```python
new_value = 90.0
year_to_update = 2020

cursor.execute(
    "UPDATE electricity_co2 SET Electricity_Access_Pct = ? WHERE Year = ?",
    (new_value, year_to_update)
)
conn.commit()

print(f"Updated {cursor.rowcount} row(s)")
```

4.7 Restored the full dataset

```python
year_to_delete = 1996

cursor.execute(
    "DELETE FROM electricity_co2 WHERE Year = ?",
    (year_to_delete,)
)
conn.commit()

print(f"Deleted {cursor.rowcount} row(s)")
```

4.8 Loaded database data back into pandas

```python
combined.to_sql("electricity_co2", conn, if_exists="replace", index=False)

result = cursor.execute("SELECT * FROM electricity_co2 ORDER BY Year").fetchall()
print(f"Total rows: {len(result)}")
```

 4.9 Ran a filtered query

```python
from_db = pd.read_sql_query("SELECT * FROM electricity_co2 ORDER BY Year", conn)
from_db
```

```python
high_access = pd.read_sql_query(
    "SELECT * FROM electricity_co2 WHERE Electricity_Access_Pct > 80 ORDER BY Year",
    conn
)
high_access
```

```python
conn.close()
```

```python
import subprocess

subprocess.check_call(['pip', 'install', 'jinja2'])

combined.style.background_gradient(subset=['CO2_Emissions_PerCapita'], cmap='Oranges')
```

```python
def highlight_high_values(row):
    styles = []
    
    if row['Electricity_Access_Pct'] > 85:
        styles.append('background-color: lightblue')
    else:
        styles.append('')
    
    if row['CO2_Emissions_PerCapita'] > 8:
        styles.append('background-color: lightcoral')
    else:
        styles.append('')
    
    return styles

combined.style.apply(highlight_high_values, axis=1, subset=['Electricity_Access_Pct', 'CO2_Emissions_PerCapita'])
```

```python
import numpy as np

elec = combined['Electricity_Access_Pct'].to_numpy()
co2 = combined['CO2_Emissions_PerCapita'].to_numpy()

summary_stats = pd.DataFrame({
    'Metric': ['Electricity_Access_Pct', 'CO2_Emissions_PerCapita'],
    'Mean': [elec.mean(), co2.mean()],
    'Median': [np.median(elec), np.median(co2)],
    'Min': [elec.min(), co2.min()],
    'Max': [elec.max(), co2.max()],
    'Std': [elec.std(), co2.std()]
})

summary_stats.style.background_gradient(
    cmap='Blues', 
    subset=['Mean', 'Median', 'Min', 'Max', 'Std']
)
```
